Example of doing a one-sample multivariate t-test to determine if a set of 3D coordinates differ from 0,0,0

Uses:
https://github.com/dionresearch/hotelling

```python
import numpy as np
from hotelling.stats import hotelling_t2
```

```python
# Generate some fake data. Data matrices are day x 3D coordinate
nDays=7
delt1=np.random.randn(nDays,3)+1 # differs from 0,0,0 by 1 stdev
delt2=np.random.randn(nDays,3) # does NOT differ from 0,0,0
```

```python
res = hotelling_t2(delt1, np.zeros(3))
#print(round(res[0], 4)) # T2 stat
#print(round(res[1], 4)) # F stat
print('p-value when null hypothesis is false: p=%f' % round(res[2], 5)) # P-Value
```

```python
res = hotelling_t2(delt2, np.zeros(3))
print('p-value when null hypothesis is true: p=%f' % round(res[2], 5)) # P-Value
```
